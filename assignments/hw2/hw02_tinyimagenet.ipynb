## 5. Как считается оценка

Пусть $a_{pub}$ и $a_{priv}$ обозначают accuracy на Public и Private соответственно,
в долях от 0 до 1. Например, 50% в формулах записывается как 0.5.

Основная часть за качество считается по **Public**:

$$Q(a_{pub})=\begin{cases}
0,&a_{pub}<0.25,\\
\min(14,28a_{pub}),&a_{pub}\geq0.25.
\end{cases}$$

25% дают 7 баллов за качество; от 25% до 50% баллы растут линейно; 50% и выше
дают 14. Ниже 25% баллы за качество не начисляются.

Бонус за качество считается по **Private**:

$$B(a_{priv})=\max(0,28a_{priv}-14).$$

Он начинается выше 50% на Private. Бонус за первое, второе и третье места
на Private равен 5, 3 и 1 баллу соответственно; для остальных мест $P=0$.

Если $R$ обозначает баллы за отчёт (до 3), $T$ за логирование (до 3), то

$$G=\min\left(30,\left\lceil R+T+Q(a_{pub})+B(a_{priv})+P\right\rceil\right).$$

Округляется **вверх одна итоговая сумма**, не каждое слагаемое отдельно.
Общий максимум 30 включает все бонусы, в том числе за место.
Для расчёта используем значения score **выбранной финальной посылки**,
зафиксированные Kaggle, а не лучший Public-score другой посылки, который может
показываться в общей таблице. Дополнительного предварительного округления
accuracy не делаем.

Примеры при полном отчёте и логировании ($R=T=3$), без бонуса за место:

| Public | Private | До округления | Итог |
|---:|---:|---:|---:|
| 25% | 25% | 6 + 7 = 13 | 13 |
| 40% | 35% | 6 + 11.2 = 17.2 | 18 |
| 50% | 45% | 6 + 14 = 20 | 20 |
| 60% | 55% | 6 + 14 + 1.4 = 21.4 | 22 |
| 70% | 65% | 6 + 14 + 4.2 = 24.2 | 25 |
